```python
import sys
from pathlib import Path


ROOT = Path("C:/Users/Prabu/Downloads/Katabatic").resolve()

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from utils import discretize_preprocess
from katabatic.pipeline.train_test_split.pipeline import TrainTestSplitPipeline
from katabatic.models.medgan.models import MEDGAN
```

```python
input_path = ROOT / "raw_data" / "shuttle.csv"
output_path = ROOT / "discretized_data" / "shuttle.csv"

output_path.parent.mkdir(parents=True, exist_ok=True)

discretize_preprocess(str(input_path), str(output_path))
```

```python
input_csv = str(output_path)
output_dir = str(ROOT / "sample_data" / "shuttle")
synthetic_dir = str(ROOT / "synthetic" / "shuttle" / "medgan")
real_test_dir = output_dir


Path(output_dir).mkdir(parents=True, exist_ok=True)
Path(synthetic_dir).mkdir(parents=True, exist_ok=True)


pipeline = TrainTestSplitPipeline(model=MEDGAN)

result = pipeline.run(
    input_csv=input_csv,
    output_dir=output_dir,
    synthetic_dir=synthetic_dir,
    real_test_dir=real_test_dir,
)

print(result)
```
