```python


from pathlib import Path
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

from katabatic.evaluate.tstr.evaluation import TSTREvaluation


ROOT = Path("/content/drive/MyDrive/katabatic1").resolve()

DATA_DIR = ROOT / "sample_data" / "car"
RESULTS_DIR = ROOT / "Results" / "car"
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

SYNTH_DIR = DATA_DIR / "tabebm_synth"

print("Using synthetic:", SYNTH_DIR)
print("Using real test:", DATA_DIR)


X_synth = pd.read_csv(SYNTH_DIR / "x_synth.csv")
y_synth = pd.read_csv(SYNTH_DIR / "y_synth.csv").iloc[:, 0]

X_test = pd.read_csv(DATA_DIR / "x_test.csv")
y_test = pd.read_csv(DATA_DIR / "y_test.csv").iloc[:, 0]


encoder = OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)

X_all = pd.concat([X_synth, X_test], axis=0)
X_all_enc = encoder.fit_transform(X_all)

X_synth_enc = X_all_enc[:len(X_synth)]
X_test_enc  = X_all_enc[len(X_synth):]


ENC_DIR = DATA_DIR / "tabebm_encoded"
ENC_DIR.mkdir(exist_ok=True)

pd.DataFrame(X_synth_enc, columns=X_synth.columns).to_csv(ENC_DIR / "x_synth.csv", index=False)
pd.DataFrame(y_synth).to_csv(ENC_DIR / "y_synth.csv", index=False)

pd.DataFrame(X_test_enc, columns=X_test.columns).to_csv(ENC_DIR / "x_test.csv", index=False)
pd.DataFrame(y_test).to_csv(ENC_DIR / "y_test.csv", index=False)


evaluator = TSTREvaluation(
    synthetic_dir=str(ENC_DIR),
    real_test_dir=str(ENC_DIR),
    dataset_name="car",
    model_name="tabebm",
    output_dir=str(RESULTS_DIR),
)

results_df = evaluator.evaluate()

print("\n TSTR RESULTS — CAR (TabEBM)")
display(results_df)


import pandas as pd

results_df = pd.DataFrame(results_df).T
results_df.index.name = "Model"
results_df = results_df.reset_index()

out_csv = RESULTS_DIR / "tabebm_tstr.csv"
results_df.to_csv(out_csv, index=False)

print(f"Saved results to: {out_csv}")
display(results_df)
```

```python
from pathlib import Path
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

from katabatic.pipeline.train_test_split.pipeline import TrainTestSplitPipeline
from katabatic.models.tabebm.adapter import TabEBMAdapter
from katabatic.evaluate.tstr.evaluation import TSTREvaluation



ROOT = Path("/content/drive/MyDrive/katabatic1").resolve()

RAW_CSV     = ROOT / "raw_data" / "adult.csv"
DATA_DIR    = ROOT / "synthetic" / "adult"
RESULTS_DIR = ROOT / "Results" / "adult"

RESULTS_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", ROOT)
print("Running ADULT TabEBM pipeline...")



pipeline = TrainTestSplitPipeline(
    model=TabEBMAdapter
)

pipeline.run(
    input_csv=str(RAW_CSV),
    output_dir=str(DATA_DIR),
)

print("ADULT TabEBM synthetic generation finished.")



SYNTH_CSV = DATA_DIR / "tabebm.csv"

df_synth = pd.read_csv(SYNTH_CSV)

X_test = pd.read_csv(DATA_DIR / "x_test.csv")
y_test = pd.read_csv(DATA_DIR / "y_test.csv").iloc[:, 0]

# detect label column safely
possible_labels = set(df_synth.columns) - set(X_test.columns)

assert len(possible_labels) == 1, f"Expected 1 label column, found {possible_labels}"

label_col = possible_labels.pop()
print("Detected label column:", label_col)

X_synth = df_synth.drop(columns=[label_col])
y_synth = df_synth[label_col]


print("Synthetic samples:", X_synth.shape)
print("Real test samples:", X_test.shape)



X_synth = X_synth[X_test.columns]



encoder = OrdinalEncoder(
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

X_all = pd.concat([X_synth, X_test], axis=0)
X_all_enc = encoder.fit_transform(X_all)

X_synth_enc = X_all_enc[:len(X_synth)]
X_test_enc  = X_all_enc[len(X_synth):]


ENC_DIR = DATA_DIR / "tabebm_encoded"
ENC_DIR.mkdir(exist_ok=True)

enc_cols = [f"f{i}" for i in range(X_synth_enc.shape[1])]

pd.DataFrame(X_synth_enc, columns=enc_cols).to_csv(
    ENC_DIR / "x_synth.csv", index=False
)
pd.DataFrame(y_synth).to_csv(
    ENC_DIR / "y_synth.csv", index=False
)

pd.DataFrame(X_test_enc, columns=enc_cols).to_csv(
    ENC_DIR / "x_test.csv", index=False
)
pd.DataFrame(y_test).to_csv(
    ENC_DIR / "y_test.csv", index=False
)

print("Encoded data prepared.")


evaluator = TSTREvaluation(
    synthetic_dir=str(ENC_DIR),
    real_test_dir=str(ENC_DIR),
    dataset_name="adult",
    model_name="tabebm",
    output_dir=str(RESULTS_DIR),
)

results = evaluator.evaluate()

print("\nTSTR RESULTS — ADULT (TabEBM)")
print(results)



results_df = pd.DataFrame(results).T
results_df.index.name = "Model"
results_df = results_df.reset_index()

out_csv = RESULTS_DIR / "tabebm_tstr.csv"
results_df.to_csv(out_csv, index=False)

print(f"Saved results to: {out_csv}")
```
