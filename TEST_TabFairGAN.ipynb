```python
import torch
import pandas as pd
from katabatic.pipeline.train_test_split.pipeline import TrainTestSplitPipeline
from katabatic.models.tabfairgan.adapter import KatabaticTabFairGAN
from utils import discretize_preprocess

# --- Configuration ---
DATASET = "adult"
raw_path = f"raw_data/{DATASET}.csv"
discretized_path = f"discretized_data/{DATASET}.csv"
output_dir = f"sample_data/{DATASET}"
synthetic_dir = f"synthetic/{DATASET}/tabfairgan"
real_test_dir = f"sample_data/{DATASET}"

# Device Config
device = "cuda:0" if torch.cuda.is_available() else "cpu"

# Updated Configuration in your main script
model_config = {
    'epochs': 50,
    'batch_size': 256,
    'device': device,
    'fairness_config': {
        'fair_epochs': 10,
        'lamda': 0.5,
        'S': 'sex',
        'Y': 'class',      # NOTE: Your data header is 'class', not 'income'
        
        # CHANGED: Use the encoded values (strings of ints), not the original labels
        'S_under': '0',    # 0 = ' Female' (alphabetically first)
        'Y_desire': '1'    # 1 = ' >50K' (alphabetically second)
    }
}

# --- 1. Preprocess ---
print("Discretizing data...")
discretize_preprocess(
    file_path=raw_path,
    output_path=discretized_path,
    bins=10,
    strategy='uniform'
)

# --- 2. Run Pipeline ---
# We pass the class 'KatabaticTabFairGAN' (not an instance).
# The pipeline will instantiate it: model = KatabaticTabFairGAN(**config)
pipeline = TrainTestSplitPipeline(model=KatabaticTabFairGAN)

print(f"Starting pipeline for TabFairGAN on {DATASET}")
pipeline.run(
    input_csv=discretized_path,
    output_dir=output_dir,
    synthetic_dir=synthetic_dir,
    real_test_dir=real_test_dir,
    fairness_eval = True,
    **model_config
)
```
